```python
import sys
from pathlib import Path

def add_suffix_to_filenames(directory_path, suffix=""):
    """
    Adds a specified suffix to the stem of each file's name in a directory.

    Args:
        directory_path (str or Path): The path to the directory containing the files.
        suffix (str): The suffix to add to the filename stem (default is "_1").
    """
    target_directory = Path(directory_path)

    if not target_directory.is_dir():
        print(f"Error: Directory not found at {target_directory}", file=sys.stderr)
        return # Exit the function

    print(f"Adding '{suffix}' to files in: {target_directory}")

    for file_path in target_directory.iterdir():
        if file_path.is_file():
            old_name = file_path.name
            # Construct the new name: stem + suffix + original_suffix
            new_name = file_path.stem + suffix + file_path.suffix
            new_file_path = file_path.with_name(new_name)

            # Avoid renaming if the new name is the same (unlikely with adding)
            if file_path != new_file_path:
                try:
                    file_path.rename(new_file_path)
                    print(f"Renamed '{old_name}' to '{new_name}'")
                except OSError as e:
                    print(f"Error renaming '{old_name}': {e}", file=sys.stderr)
                except Exception as e:
                    print(f"An unexpected error occurred while renaming '{old_name}': {e}", file=sys.stderr)
            else:
                 print(f"Skipping '{old_name}': New name is the same as the old name.")

    print("Adding suffix process completed.")


```

```python
def remove_suffix_from_filenames(directory_path, suffix="_1"):
    """
    Removes a specified suffix from the end of the stem of each file's name
    in a directory, if the stem ends with that suffix.

    Args:
        directory_path (str or Path): The path to the directory containing the files.
        suffix (str): The suffix to remove from the filename stem (default is "_1").
    """
    target_directory = Path(directory_path)

    if not target_directory.is_dir():
        print(f"Error: Directory not found at {target_directory}", file=sys.stderr)
        return # Exit the function

    print(f"Checking files in: {target_directory} to remove '{suffix}'")

    for file_path in target_directory.iterdir():
        if file_path.is_file():
            old_name = file_path.name
            old_stem = file_path.stem # Filename without extension

            # Check if the stem ends with the specified suffix
            if old_stem.endswith(suffix):
                # Construct the new stem by removing the suffix
                # [:-len(suffix)] slices the string to remove the last 'len(suffix)' characters
                new_stem = old_stem[:-len(suffix)]

                # Construct the new full filename
                new_name = new_stem + file_path.suffix
                new_file_path = file_path.with_name(new_name)

                # Avoid trying to rename to the exact same name
                if file_path != new_file_path:
                    try:
                        file_path.rename(new_file_path)
                        print(f"Renamed '{old_name}' to '{new_name}'")
                    except OSError as e:
                        print(f"Error renaming '{old_name}': {e}", file=sys.stderr)
                    except Exception as e:
                        print(f"An unexpected error occurred while renaming '{old_name}': {e}", file=sys.stderr)
                else:
                     # This else block is mostly for safety/clarity
                    print(f"Skipping '{old_name}': New name is the same as the old name.")
            # else: # Optional: uncomment if you want to see files that don't match
            #     print(f"'{old_name}' does not end with '{suffix}', skipping.")


    print("Removing suffix process completed.")
```
